```python
# -*- coding: utf-8 -*-
"""
极端降雨群发滑坡智能识别研究 - 发文量趋势分析
基于真实 WoS 数据 (savedrecs.xls)
完整绘图代码
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import gridspec
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from collections import defaultdict, Counter
import math
import re

import seaborn as sns
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import spacy
import gensim
from gensim.corpora import Dictionary
from gensim.models import LdaMulticore, LdaModel, CoherenceModel
from gensim.matutils import Sparse2Corpus
import warnings
import os
import logging
import nltk
nltk.download('stopwords')

```

```python
# ==================== 2. 加载真实数据（基于 Early Access Date） ====================
df_raw = pd.read_excel('/content/drive/MyDrive/savedrecs.xls', engine='xlrd')

date_col = 'Early Access Date'
if date_col not in df_raw.columns:
    raise KeyError(f"未找到 '{date_col}' 列，请检查数据列名。")

def extract_year(date_str):
    if pd.isna(date_str):
        return np.nan
    s = str(date_str).strip()
    if not s:
        return np.nan
    parts = s.split()
    try:
        return int(parts[-1])
    except ValueError:
        try:
            return int(s)
        except ValueError:
            return np.nan

df_raw['year_extracted'] = df_raw[date_col].apply(extract_year)

year_series = df_raw['year_extracted'].dropna().astype(int)
year_counts = year_series.value_counts().sort_index()

all_years = np.arange(year_counts.index.min(), year_counts.index.max() + 1)
counts = np.array([year_counts.get(y, 0) for y in all_years])

df = pd.DataFrame({'year': all_years, 'count': counts})
df['cumulative'] = df['count'].cumsum()

print(f"数据年份范围：{df['year'].min()} - {df['year'].max()}")
print(f"总发文量：{df['count'].sum()} 篇")
print(f"总年份数：{len(df)} 年")
```

```python
# ==================== 3. 阶段划分（根据实际趋势手动调整） ====================
stages = [
    ('Emergence Stage', 2002, 2012),
    ('Steady Growth Stage', 2013, 2019),
    ('Rapid Growth Stage', 2020, 2025)
]
```

```python
# ==================== 4. 重大极端降雨事件（根据事实标注） ====================
events = [
    (2021, 'Henan "7·20" Rainstorm'),
    (2022, 'Guangxi Longsheng Landslide'),
    (2023, 'Typhoon Haikui'),
    (2024, 'Chongqing Pengshui Rainstorm'),

    # 新增重大事件（建议选取 5-6 个，请按需调整）
    (2013, 'Sichuan Lushan Rain-induced Slide'),  # 四川芦山地震后强降雨引发滑坡（可选）
    (2017, 'Sichuan Maoxian Landslide'),          # 四川茂县叠溪镇山体滑坡（2017.6.24））
]
```

```python
# ==================== 5. 多项式拟合（累计发文量） ====================
X = df['year'].values.reshape(-1, 1)
y = df['cumulative'].values
poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)
model = LinearRegression()
model.fit(X_poly, y)
y_pred = model.predict(X_poly)
r2 = r2_score(y, y_pred)

# 提取系数用于显示方程
coeffs = model.coef_
intercept = model.intercept_

```

```python
# ==================== 6. 绘制组合图 ====================
fig = plt.figure(figsize=(14, 10))
gs = gridspec.GridSpec(2, 1, height_ratios=[3, 2], hspace=0.35)

# ---------- 子图1：年度发文量 ----------
ax1 = fig.add_subplot(gs[0])

# 主折线
ax1.plot(df['year'], df['count'], 'o-', color='#2c3e50',
         linewidth=2.5, markersize=7, label='Annual Publications', zorder=2)

# 阶段划分（垂直虚线 + 文本标签）
stage_colors = ['#95a5a6', '#3498db', '#e67e22']
for i, (name, start, end) in enumerate(stages):
    mid_year = (start + end) // 2
    ax1.axvline(x=start, color=stage_colors[i], linestyle='--',
                alpha=0.6, linewidth=1.5, zorder=1)
    # 阶段名称标注在图表顶部
    ax1.text(mid_year, ax1.get_ylim()[1] * 0.93, name,
             ha='center', fontsize=13, color=stage_colors[i],
             weight='bold', zorder=3)

# 重大事件标注（红色三角在顶部，并旋转文本）
for year, label in events:
    if year in df['year'].values:
        # 获取该年的发文量
        count_val = df.loc[df['year'] == year, 'count'].iloc[0]
        # 在数据点位置画红色星号或三角形（更醒目）
        ax1.scatter(year, count_val, marker='^', s=120,
                    color='#c0392b', zorder=3, edgecolor='white', linewidth=1)
        # 在点上方加英文标签（向上偏移一点避免重叠）
        ax1.text(year, count_val + 2, label,
                 ha='center', va='bottom', fontsize=8,
                 color='#c0392b', weight='bold', zorder=3)

# 坐标轴设置
ax1.set_ylabel('Number of Publications', fontsize=13)
ax1.set_xlim(df['year'].min() - 2, df['year'].max() + 2)
ax1.set_title('Annual Publication Trends in Extreme Rainfall-Induced Landslide Research', fontsize=15, weight='bold')
ax1.grid(True, linestyle=':', alpha=0.5, axis='y')
ax1.legend(loc='lower right', fontsize=11)
ax1.tick_params(labelsize=11)
# （可选）在折线上显示数值，若年份太多可注释掉
# for x, y_val in zip(df['year'], df['count']):
#     if y_val > 0:
#         ax1.text(x, y_val + 2, str(y_val), ha='center', fontsize=8)

# ---------- 子图2：累计发文量与拟合曲线 ----------
ax2 = fig.add_subplot(gs[1])

# 实际累计曲线
ax2.plot(df['year'], df['cumulative'], 'o-', color='#2c3e50',
         linewidth=2, markersize=5, label='Cumulative Publications', zorder=2)

# 三阶多项式拟合曲线（红色虚线）
ax2.plot(df['year'], y_pred, '--', color='#e74c3c',
         linewidth=2.5, label=f'Cubic Polynomial Fit (R² = {r2:.4f})', zorder=3)

# 显示拟合方程
eq_text = (f'y = {coeffs[3]:.3f}x³ + {coeffs[2]:.3f}x² + '
           f'{coeffs[1]:.3f}x + {intercept:.0f}')
ax2.text(0.05, 0.92, eq_text, transform=ax2.transAxes,
         fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='#fef9e7', edgecolor='#e74c3c', alpha=0.9))

# 坐标轴设置
ax2.set_xlabel('Year', fontsize=13)
ax2.set_ylabel('Cumulative Publications', fontsize=13)
ax2.set_xlim(df['year'].min() - 2, df['year'].max() + 2)
ax2.grid(True, linestyle=':', alpha=0.5, axis='y')
ax2.legend(loc='lower right', fontsize=11)
ax2.tick_params(labelsize=11)
```

#作者分析

```python
# ------------------------------
# 1. 读取数据
# ------------------------------
file_path = '/content/drive/MyDrive/savedrecs.xls'  # 请根据实际路径修改
df = pd.read_excel(file_path, sheet_name=0)  # 读取第一个sheet

# 检查必要的列是否存在
required_cols = ['Authors', 'Times Cited, WoS Core']
for col in required_cols:
    if col not in df.columns:
        raise ValueError(f"缺少列: {col}")

# 处理被引次数，转为数值，缺失值填0
df['Times Cited, WoS Core'] = pd.to_numeric(df['Times Cited, WoS Core'], errors='coerce').fillna(0)

# 只保留有作者的行
df = df[df['Authors'].notna()]
```

```python
# ------------------------------
# 2. 解析作者名单（支持分号分隔）
# ------------------------------
# 作者格式示例: "Amankwah, SOY; Wang, GJ; Gnyawali, K; ..."
# 为了统一，去除多余空格，并拆分为列表
df['Authors_parsed'] = df['Authors'].apply(lambda x: [name.strip() for name in x.split(';') if name.strip()])
```

```python
# ------------------------------
# 3. 构建作者-论文映射
# ------------------------------
# 记录每篇文章的被引次数（用于后续计算）
papers = df[['Authors_parsed', 'Times Cited, WoS Core']].copy()
```

```python
# ------------------------------
# 4. 计算作者指标
# ------------------------------
# 4.1 总发文量 & 总被引
author_pub_count = defaultdict(int)
author_citation_sum = defaultdict(float)

for idx, row in papers.iterrows():
    authors = row['Authors_parsed']
    citations = row['Times Cited, WoS Core']
    for author in authors:
        author_pub_count[author] += 1
        author_citation_sum[author] += citations

# 4.2 篇均被引
author_cited_per_pub = {a: author_citation_sum[a] / author_pub_count[a] for a in author_pub_count}

# 4.3 H指数
def compute_h_index(citation_list):
    citation_list = sorted([c for c in citation_list if c > 0], reverse=True)
    h = 0
    for i, c in enumerate(citation_list, 1):
        if c >= i:
            h = i
        else:
            break
    return h

# 收集每位作者的所有单篇被引次数
author_citations_list = defaultdict(list)
for idx, row in papers.iterrows():
    citations = row['Times Cited, WoS Core']
    for author in row['Authors_parsed']:
        if citations > 0:
            author_citations_list[author].append(citations)

author_h_index = {a: compute_h_index(author_citations_list[a]) for a in author_pub_count}

# 4.4 合作分析
# 构建共现矩阵：统计每对作者合作次数
co_author_counts = defaultdict(Counter)

for idx, row in papers.iterrows():
    authors = row['Authors_parsed']
    for i, a1 in enumerate(authors):
        for a2 in authors[i+1:]:
            if a1 != a2:
                co_author_counts[a1][a2] += 1
                co_author_counts[a2][a1] += 1

# 4.5 最紧密合作对象 & 合作总数
most_frequent_coauthor = {}
total_coauthors = {}

for author in author_pub_count.keys():
    if author in co_author_counts:
        # 合作总数：不重复合作者数量
        total_coauthors[author] = len(co_author_counts[author])
        # 最紧密合作对象（合作次数最多）
        if co_author_counts[author]:
            most_freq = max(co_author_counts[author].items(), key=lambda x: x[1])
            most_frequent_coauthor[author] = most_freq[0]  # 只存名字，也可存次数
        else:
            most_frequent_coauthor[author] = None
    else:
        total_coauthors[author] = 0
        most_frequent_coauthor[author] = None
```

```python

# ------------------------------
# 5. 构建作者信息DataFrame
# ------------------------------
author_df = pd.DataFrame({
    'Author': list(author_pub_count.keys()),
    '总发文量': list(author_pub_count.values()),
    '总被引': list(author_citation_sum.values()),
    '篇均被引': [author_cited_per_pub[a] for a in author_pub_count.keys()],
    'H指数': [author_h_index[a] for a in author_pub_count.keys()],
    '合作总数': [total_coauthors.get(a, 0) for a in author_pub_count.keys()],
    '最紧密合作对象': [most_frequent_coauthor.get(a, '') for a in author_pub_count.keys()]
})

# 按总发文量降序排序
author_df_sorted = author_df.sort_values('总发文量', ascending=False).reset_index(drop=True)
```

```python
# ------------------------------
# 6. 输出结果
# ------------------------------
print("核心作者排名表（按总发文量）：")
print(author_df_sorted.to_string(index=False))

# 保存为Excel文件
author_df_sorted.to_excel("核心作者分析结果.xlsx", index=False)
```

#机构分析

```python
file_path = '/content/drive/MyDrive/savedrecs.xls'  # 请根据实际路径修改
df = pd.read_excel(file_path, sheet_name=0)  # 读取第一个sheet

# 检查必要列是否存在
required_cols = ['Authors', 'Affiliations', 'Times Cited, WoS Core']
for col in required_cols:
    if col not in df.columns:
        raise KeyError(f"列 '{col}' 不存在，请检查数据列名。")

# 处理缺失值
df['Authors'] = df['Authors'].fillna('')
df['Affiliations'] = df['Affiliations'].fillna('')
df['Times Cited, WoS Core'] = df['Times Cited, WoS Core'].fillna(0).astype(int)

# 定义机构合并函数（中国科学院）
def normalize_institution(inst):
    """
    将机构名称标准化，合并中国科学院及其子机构为 'Chinese Academy of Sciences'
    """
    if not isinstance(inst, str):
        return inst
    # 1. 合并中国科学院及子机构
    if any(key in inst for key in ['Chinese Acad Sci', 'Chinese Academy of Sciences', 'Institute of Mountain Hazards', 'Inst Mt Hazards']):
        return 'Chinese Academy of Sciences'
    # 2. 合并中国地质科学院及子机构
    if any(key in inst for key in ['Chinese Acad Geol Sci', 'Chinese Academy of Geological Sciences', 'Institute of Geomechanics']):
        return 'Chinese Academy of Geological Sciences'
    # 3. 合并印度理工学院及所有分校
    if 'Indian Institute of Technology' in inst:
        return 'Indian Institute of Technology System (IIT System)'
    # 4. 合并印度空间研究组织与空间部
    if 'Indian Space Research Organisation' in inst or 'Department of Space' in inst:
        return 'Indian Space Research Organisation (ISRO)'

    return inst

# 解析作者和机构列表
def parse_authors(authors_str):
    """解析作者字符串，返回作者列表"""
    if pd.isna(authors_str) or authors_str == '':
        return []
    # 按分号分割，并去除前后空格
    authors = [a.strip() for a in authors_str.split(';') if a.strip()]
    return authors

def parse_affiliations(aff_str):
    """解析机构字符串，返回机构列表（标准化后）"""
    if pd.isna(aff_str) or aff_str == '':
        return []
    # 按分号分割
    insts = [i.strip() for i in aff_str.split(';') if i.strip()]
    # 标准化每个机构
    insts = [normalize_institution(i) for i in insts]
    return insts

# 用于存储每个机构的论文信息
paper_info = defaultdict(list)  # 机构 -> [(times_cited, authors_list, insts_list)]

# 逐行处理
for idx, row in df.iterrows():
    authors_str = row['Authors']
    aff_str = row['Affiliations']
    cited = row['Times Cited, WoS Core']

    authors = parse_authors(authors_str)
    insts = parse_affiliations(aff_str)

    if not insts:  # 无机构信息则跳过
        continue

    # 对每个机构记录该论文的信息
    # 注意：一篇论文可能对应多个机构，每个机构都获得该论文的贡献
    for inst in set(insts):  # 去重，避免同一机构重复计数
        paper_info[inst].append({
            'cited': cited,
            'authors': authors,
            'insts': insts  # 用于合作机构分析
        })

# 计算各机构的统计指标
results = []
for inst, papers in paper_info.items():
    total_pubs = len(papers)
    total_citations = sum(p['cited'] for p in papers)
    avg_citations = total_citations / total_pubs if total_pubs > 0 else 0

    # H指数
    citations_list = sorted([p['cited'] for p in papers], reverse=True)
    h_index = 0
    for i, c in enumerate(citations_list, 1):
        if c >= i:
            h_index = i
        else:
            break

    # 合作机构总数：统计该机构所有论文中出现的其他机构（去重）
    co_insts = set()
    for p in papers:
        for other_inst in p['insts']:
            if other_inst != inst:
                co_insts.add(other_inst)
    co_count = len(co_insts)

    # 研究人员数：统计该机构所有论文中的所有作者（去重）
    # 这里采用简单方法：将论文的所有作者都计入该机构，但可能导致作者跨机构重复，但作为近似
    # 更精确应基于作者-机构对应关系，但此处简化。
    researchers = set()
    for p in papers:
        for author in p['authors']:
            researchers.add(author)
    researcher_count = len(researchers)

    # 人均产出
    per_capita = total_pubs / researcher_count if researcher_count > 0 else 0

    results.append({
        '机构': inst,
        '总发文量': total_pubs,
        '总被引': total_citations,
        '篇均被引': round(avg_citations, 2),
        'H指数': h_index,
        '合作机构总数': co_count,
        '研究人员数': researcher_count,
        '人均产出': round(per_capita, 3)
    })

# 转换为DataFrame并排序（按总发文量降序）
df_results = pd.DataFrame(results)
df_results = df_results.sort_values('总发文量', ascending=False).reset_index(drop=True)

# 输出前20名核心机构
print("核心机构排名（按发文量）")
print(df_results.head(20).to_string(index=False))

# 可保存为Excel
df_results.to_excel('institution_analysis.xlsx', index=False)
```

#国家分析

```python

file_path = '/content/drive/MyDrive/savedrecs.xls'  # 请根据实际路径修改
df = pd.read_excel(file_path, sheet_name=0)  # 读取第一个sheet

# 国家名称映射
country_mapping = {
    'Peoples R China': 'China',
    'USA': 'United States',
    'England': 'UK',
    'Scotland': 'UK',
    'Wales': 'UK',
    'South Korea': 'Korea',
    # 根据需要补充
}

def extract_countries(addr_str):
    """从地址字符串中提取所有出现的不重复国家"""
    if pd.isna(addr_str):
        return []
    countries = set()
    # 以分号分隔各个地址块
    for addr in addr_str.split(';'):
        # 取最后一个逗号后的部分作为国家
        parts = addr.split(',')
        if len(parts) > 1:
            country = parts[-1].strip()
            # 简单清洗：去除末尾句号或括号等
            country = re.sub(r'[\.\)].*$', '', country).strip()
            if country in country_mapping:
                country = country_mapping[country]
            if country:  # 非空
                countries.add(country)
    return list(countries)

# 提取每篇论文的国家列表
df['countries'] = df['Addresses'].apply(extract_countries)

# 删除无国家信息的记录（可选）
df = df[df['countries'].map(len) > 0]

# 获取被引次数（若缺失则按0计）
df['cites'] = pd.to_numeric(df['Times Cited, WoS Core'], errors='coerce').fillna(0)

# ---------- 国家统计 ----------
nation_papers = defaultdict(list)  # 国家 -> 该国家所有论文的被引次数列表

for idx, row in df.iterrows():
    for c in row['countries']:
        nation_papers[c].append(row['cites'])

stats = {}
for c, cites_list in nation_papers.items():
    n = len(cites_list)
    total_cites = sum(cites_list)
    avg_cites = total_cites / n if n > 0 else 0
    # 计算H指数
    cites_sorted = sorted(cites_list, reverse=True)
    h = 0
    for i, c_val in enumerate(cites_sorted, 1):
        if c_val >= i:
            h = i
        else:
            break
    stats[c] = {
        'papers': n,
        'total_cites': total_cites,
        'avg_cites': round(avg_cites, 2),
        'h_index': h
    }

# ---------- 合作网络 ----------
coop = defaultdict(Counter)  # 国家A -> Counter({国家B: 合作次数})
for idx, row in df.iterrows():
    cs = row['countries']
    if len(cs) > 1:
        for i in range(len(cs)):
            for j in range(i+1, len(cs)):
                coop[cs[i]][cs[j]] += 1
                coop[cs[j]][cs[i]] += 1

# 最紧密合作对象
most_coop = {}
for c in stats:
    if c in coop:
        most = coop[c].most_common(1)
        if most:
            most_coop[c] = most[0][0]

# 输出结果（按发文量降序）
print("国家\t发文量\t总被引\t篇均被引\tH指数\t最紧密合作对象\t合作国家数")
for c in sorted(stats, key=lambda x: stats[x]['papers'], reverse=True):
    data = stats[c]
    partner = most_coop.get(c, '无')
    num_partners = len(coop.get(c, {}))
    print(f"{c}\t{data['papers']}\t{data['total_cites']}\t{data['avg_cites']}\t{data['h_index']}\t{partner}\t{num_partners}")
```

```python
file_path = '/content/drive/MyDrive/savedrecs.xls'  # 请根据实际路径修改
df = pd.read_excel(file_path, sheet_name=0)  # 读取第一个sheet

# 查看列名（按需调整）
print(df.columns.tolist())

# 选取所需字段
journal_col = 'Source Title'
cited_col = 'Times Cited, WoS Core'  # 若无则使用 'Times Cited, All Databases'
doc_type_col = 'Document Type'

# 清理数据：过滤掉无期刊名或引用数为空的记录
df_clean = df[[journal_col, cited_col, doc_type_col]].dropna(subset=[journal_col])
# 将引用数转为数值，无效值填充为0
df_clean[cited_col] = pd.to_numeric(df_clean[cited_col], errors='coerce').fillna(0)

# 若只考虑期刊论文，可过滤 Document Type 为 'Article' 或 'Review'
# df_clean = df_clean[df_clean[doc_type_col].isin(['Article', 'Review'])]

print(f"有效记录数：{len(df_clean)}")
```

##统计指标计算

```python
def compute_h_index(citations_list):
    """计算H指数"""
    citations = sorted(citations_list, reverse=True)
    h = 0
    for i, c in enumerate(citations, start=1):
        if c >= i:
            h = i
        else:
            break
    return h

# 按期刊分组
grouped = df_clean.groupby(journal_col)

journal_stats = {}
for journal, group in grouped:
    n = len(group)
    total_cites = group[cited_col].sum()
    avg_cites = total_cites / n if n > 0 else 0
    h_index = compute_h_index(group[cited_col].tolist())
    journal_stats[journal] = {
        '发文量': n,
        '总被引': total_cites,
        '篇均被引': avg_cites,
        'H指数': h_index
    }

# 转为DataFrame便于排序
stats_df = pd.DataFrame.from_dict(journal_stats, orient='index')
stats_df.index.name = '期刊'
stats_df.reset_index(inplace=True)

# 按总发文量降序排序
stats_df_sorted = stats_df.sort_values('发文量', ascending=False)
```

##核心期刊群识别

```python
stats_df_sorted['累积发文量'] = stats_df_sorted['发文量'].cumsum()
total_papers = stats_df_sorted['发文量'].sum()
stats_df_sorted['累积占比'] = stats_df_sorted['累积发文量'] / total_papers * 100

# 识别核心期刊（累积占比 ≤ 80%）
core_mask = stats_df_sorted['累积占比'] <= 80
core_num = core_mask.sum()  # 核心期刊数量

# 获取前20名期刊
top20 = stats_df_sorted.head(20).copy()
top20['排名'] = range(1, len(top20) + 1)
top20['是否核心'] = top20['累积占比'] <= 80

# 输出前20名期刊排名表（含核心标记）
print("\n=== 前20名期刊排名（按发文量） ===\n")
print(top20[['排名', '期刊', '发文量', '总被引', '篇均被引', 'H指数',
             '累积发文量', '累积占比', '是否核心']].to_string(index=False, float_format='%.2f'))

# 输出核心期刊总数
print(f"\n核心期刊数量（累积发文量占比 ≤ 80%）：{core_num}")
```

##基金分析

```python
# 1. 读取数据（假设文件为制表符分隔的文本，跳过可能的元数据行）
#    实际文件可能为.xls，但内容为制表符分隔，故用read_csv处理
file_path = '/content/drive/MyDrive/savedrecs.xls'  # 请根据实际路径修改
df = pd.read_excel(file_path, sheet_name=0)  # 读取第一个sheet
print(f"总记录数: {len(df)}")
```

```python
# 2. 筛选有效记录（需有基金信息）
df_funded = df[df['Funding Orgs'].notna()].copy()
print(f"有基金资助的记录数: {len(df_funded)}")
```

```python
# 3. 提取国家/地区（从通讯作者地址）
def extract_country(reprint_addr):
    if pd.isna(reprint_addr):
        return None
    # 示例: "Wang, GJ (通讯作者)，Nanjing Univ Informat Sci & Technol, ... China."
    # 取最后一个逗号后的词，去掉句号
    parts = reprint_addr.split(',')
    if len(parts) > 1:
        last = parts[-1].strip().rstrip('.')
        # 可能包含"Peoples R China"等，统一处理
        if 'China' in last:
            return 'China'
        elif 'USA' in last or 'United States' in last:
            return 'USA'
        elif 'England' in last or 'UK' in last or 'United Kingdom' in last:
            return 'UK'
        elif 'Germany' in last:
            return 'Germany'
        elif 'Japan' in last:
            return 'Japan'
        else:
            return last  # 保留原样
    return None

df_funded['Country'] = df_funded['Reprint Addresses'].apply(extract_country)
```

```python
# 4. 解析基金名称（多个基金以分号分隔）
def parse_funding(funding_str):
    if pd.isna(funding_str):
        return []
    # 按分号分隔，去除首尾空格
    funds = [f.strip() for f in funding_str.split(';') if f.strip()]
    return funds

df_funded['Funds'] = df_funded['Funding Orgs'].apply(parse_funding)
```

```python
# 5. 展开数据（每行对应一个基金）
fund_records = []
for idx, row in df_funded.iterrows():
    for fund in row['Funds']:
        fund_records.append({
            'Fund': fund,
            'Title': row['Article Title'],
            'Times_Cited': row['Times Cited, WoS Core'] if pd.notna(row['Times Cited, WoS Core']) else 0,
            'Authors': row['Authors'],
            'Country': row['Country']
        })
fund_df = pd.DataFrame(fund_records)
print(f"展开后的基金-文献记录数: {len(fund_df)}")
```

```python
# 6. 计算指标
# 总发文量（每个基金计数）
pub_count = fund_df.groupby('Fund').size().reset_index(name='Total_Pubs')
# 总被引
total_cite = fund_df.groupby('Fund')['Times_Cited'].sum().reset_index(name='Total_Cites')
# 篇均被引
avg_cite = fund_df.groupby('Fund')['Times_Cited'].mean().reset_index(name='Avg_Cites')
# H指数计算函数
def h_index(cites_list):
    cites_sorted = sorted(cites_list, reverse=True)
    h = 0
    for i, c in enumerate(cites_sorted, 1):
        if c >= i:
            h = i
        else:
            break
    return h

h_index_dict = {}
for fund, group in fund_df.groupby('Fund'):
    cites = group['Times_Cited'].tolist()
    h_index_dict[fund] = h_index(cites)
h_series = pd.Series(h_index_dict).reset_index(name='H_index')
h_series.columns = ['Fund', 'H_index']

# 国家/地区（每个基金取最主要国家，即出现次数最多的国家）
def get_main_country(fund):
    countries = fund_df[fund_df['Fund'] == fund]['Country'].dropna()
    if len(countries) == 0:
        return 'Unknown'
    return countries.value_counts().idxmax()

main_country = {}
for fund in fund_df['Fund'].unique():
    main_country[fund] = get_main_country(fund)
country_series = pd.Series(main_country).reset_index(name='Main_Country')
country_series.columns = ['Fund', 'Main_Country']

# 研究人数（平均每个基金的作者数，或总人数？此处取总作者数之和，但作者可能有重复，简单加总）
def total_authors(fund):
    authors_str = fund_df[fund_df['Fund'] == fund]['Authors']
    count = 0
    for s in authors_str:
        if pd.notna(s):
            count += len(s.split(';'))
    return count

author_count = {}
for fund in fund_df['Fund'].unique():
    author_count[fund] = total_authors(fund)
author_series = pd.Series(author_count).reset_index(name='Total_Authors')
author_series.columns = ['Fund', 'Total_Authors']

# 合并所有指标
result_df = pub_count.merge(total_cite, on='Fund') \
                     .merge(avg_cite, on='Fund') \
                     .merge(h_series, on='Fund') \
                     .merge(country_series, on='Fund') \
                     .merge(author_series, on='Fund')
result_df = result_df[['Fund', 'Total_Pubs', 'Total_Cites', 'Avg_Cites', 'H_index', 'Main_Country', 'Total_Authors']]
result_df.sort_values('Total_Pubs', ascending=False, inplace=True)
```

```python
# 7. 输出基金排名表（前20）
print("\n=== 基金项目排名表（按总发文量） ===")
print(result_df.head(20).to_string(index=False))
```

```python
# 8. 重点分析NSFC
nsfc_keywords = ['National Natural Science Foundation of China', 'NSFC', 'Natural Science Foundation of China']
def is_nsfc(fund):
    fund_lower = fund.lower()
    for kw in nsfc_keywords:
        if kw.lower() in fund_lower:
            return True
    return False

nsfc_funds = [f for f in result_df['Fund'] if is_nsfc(f)]
nsfc_records = fund_df[fund_df['Fund'].isin(nsfc_funds)]
nsfc_metrics = {
    '总发文量': len(nsfc_records),
    '总被引': nsfc_records['Times_Cited'].sum(),
    '篇均被引': nsfc_records['Times_Cited'].mean(),
    'H指数': h_index(nsfc_records['Times_Cited'].tolist()),
    '涉及基金数': len(nsfc_funds)
}
print("\n=== 中国国家自然科学基金(NSFC)分析 ===")
for k, v in nsfc_metrics.items():
    print(f"{k}: {v:.2f}" if isinstance(v, float) else f"{k}: {v}")
```

```python
# 9. 对比中外基金影响力差异（基于基金的主要国家/地区）
# 获取每个基金的主要国家（已在前面计算，这里直接从result_df取）
fund_main_country = result_df[['Fund', 'Main_Country']].set_index('Fund')['Main_Country'].to_dict()

def fund_is_china(fund):
    country = fund_main_country.get(fund, 'Unknown')
    # 将台湾视作中国的一部分
    return country in ['China', 'Taiwan']

fund_df['Is_China'] = fund_df['Fund'].apply(fund_is_china)
china_records = fund_df[fund_df['Is_China'] == True]
foreign_records = fund_df[fund_df['Is_China'] == False]

def summarize(records, name):
    if len(records) == 0:
        return
    cites = records['Times_Cited'].tolist()
    print(f"\n=== {name} 基金 ===")
    print(f"发文量: {len(records)}")
    print(f"总被引: {sum(cites)}")
    print(f"篇均被引: {np.mean(cites):.2f}")
    print(f"H指数: {h_index(cites)}")
    countries = records['Country'].dropna()
    if len(countries) > 0:
        top_countries = countries.value_counts().head(5)
        print("主要国家分布:", top_countries.to_dict())

summarize(china_records, "中国基金（含台湾）")
summarize(foreign_records, "国外基金")
```

#LDA分析

```python
#!/usr/bin/env python
# -*- coding: utf-8 -*-
"""
LDA 主题模型分析 - 滑坡文献数据
输入： savedrecs.xls（包含文献信息）
输出：
    - 最优主题数曲线（困惑度 & 一致性）
    - 主题-术语表（打印并保存为 CSV）
    - 各主题频率随年份变化趋势图（保存为 PNG）
    - 主题相似度矩阵及 VOSviewer 网络文件 (.net)
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import spacy
import gensim
from gensim.corpora import Dictionary
from gensim.models import LdaMulticore, LdaModel, CoherenceModel
from gensim.matutils import Sparse2Corpus
import warnings
import os
import logging

# 设置日志级别，屏蔽 gensim 的过多输出
logging.basicConfig(level=logging.ERROR)

# 下载 nltk 数据（首次运行时需取消注释）
# import nltk
# nltk.download('punkt')
# nltk.download('stopwords')

# 加载 spacy 英文模型（需先安装：python -m spacy download en_core_web_sm）
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])
# 扩展停用词
from nltk.corpus import stopwords

# ==================== 1. 数据读取与预处理 ====================
def load_and_preprocess(filepath):
    """
    读取 Excel/CSV 文件，合并文本字段，分词，构建词典和语料库。
    返回: (df, dictionary, corpus)
    """
    # ------ 1. 健壮读取文件 ------
    def read_file(fp):
        try:
            return pd.read_excel(fp)
        except Exception:
            try:
                return pd.read_csv(fp, encoding='utf-8')
            except Exception:
                try:
                    return pd.read_csv(fp, encoding='latin-1')
                except Exception:
                    return pd.read_csv(fp, sep='\t', encoding='utf-8')
    df = read_file(filepath)
    print("文件读取成功，列名：", df.columns.tolist())

    # ------ 2. 合并文本字段（根据实际列名调整）------
    # 注意：列名可能因文件格式而异，此处假设为原始列名，请根据实际调整
    text_cols = ['Article Title', 'Abstract', 'Author Keywords', 'Keywords Plus']
    # 检查列是否存在
    for col in text_cols:
        if col not in df.columns:
            print(f"警告：列 '{col}' 不存在，将跳过该列")
    df['combined'] = ''
    for col in text_cols:
        if col in df.columns:
            df['combined'] += df[col].fillna('') + ' '

    # ------ 3. 提取年份 ------
    if 'Publication Year' in df.columns:
        df['year'] = pd.to_numeric(df['Publication Year'], errors='coerce')
    else:
        # 尝试其他可能的年份列名
        year_col = next((c for c in df.columns if 'year' in c.lower()), None)
        if year_col:
            df['year'] = pd.to_numeric(df[year_col], errors='coerce')
        else:
            raise ValueError("未找到年份列，请检查数据")
    df = df.dropna(subset=['year'])
    df['year'] = df['year'].astype(int)
    extra_stopwords = {
    # 论文套话
    'study', 'paper', 'article', 'research', 'result', 'conclusion',
    'present', 'propose', 'proposed', 'show', 'shown', 'based', 'using',
    'used', 'use', 'also', 'however', 'therefore', 'may', 'can',
    # 泛化的量词/副词
    'high', 'low', 'large', 'small', 'different', 'various', 'significant',
    'important', 'main', 'major', 'new', 'recent',
    # 时间/地点通用词
    'year', 'time', 'day', 'period', 'region', 'area', 'site', 'location',
    # 泛化的名词
    'datum', 'base', 'factor', 'method', 'analysis', 'approach',
    'process', 'system', 'event', 'case', 'condition', 'characteristic',
    # 单位/格式词
    'et', 'al', 'fig', 'figure', 'table', 'eq', 'equation',
    }
    # 注意：不再包含 rainfall, trigger, flow, movement, debris, change, increase 等
    stop_words = set(stopwords.words('english')).union(extra_stopwords)
    # ------ 4. 文本预处理 ------
    def preprocess(text):
      if not isinstance(text, str):
          return []
      doc = nlp(text)
      tokens = [
          token.lemma_.lower()
          for token in doc
          if token.is_alpha
          and token.lemma_.lower() not in stop_words
          and token.pos_ in ('NOUN', 'ADJ', 'VERB')  # 恢复形容词和动词
          and len(token.lemma_) > 2  # 过滤过短的词
      ]
      return tokens

    print("正在预处理文本...")
    df['tokens'] = df['combined'].apply(preprocess)

    # 构建词典和语料库
    dictionary = Dictionary(df['tokens'])
    dictionary.filter_extremes(no_below=3, no_above=0.5)  # 从 0.6 降到 0.5，剔除更普遍的词
    corpus = [dictionary.doc2bow(tokens) for tokens in df['tokens']]

    return df, dictionary, corpus


def compute_coherence_values(corpus, dictionary, texts, start=2, end=15, step=1):
    """
    计算不同主题数 K 下的 Coherence (c_v) 和 Perplexity。
    返回各 K 下的 coherence 和 perplexity 列表。
    """
    coherence_scores = []
    perplexity_scores = []
    models = {}
    for k in range(start, end+1, step):
        print(f"训练 LDA K={k} ...")
        # 训练 LDA 模型，设定随机种子保证可重复
        lda = LdaModel(corpus=corpus, id2word=dictionary, num_topics=k,
                       passes=10, random_state=42, alpha='auto', eta='auto')
        models[k] = lda
        # 计算一致性
        cm = CoherenceModel(model=lda, texts=texts, dictionary=dictionary, coherence='c_v')
        coh = cm.get_coherence()
        # 计算困惑度
        perp = lda.log_perplexity(corpus)
        coherence_scores.append(coh)
        perplexity_scores.append(perp)
        print(f"K={k}   Coherence={coh:.4f}   Perplexity={perp:.4f}")
    return models, coherence_scores, perplexity_scores


def plot_optimal_k(coherence_scores, perplexity_scores, start_k, end_k):
    """
    绘制一致性（越高越好）和困惑度（越低越好）随 K 的变化曲线，辅助确定最优主题数。
    """
    ks = list(range(start_k, end_k+1))
    fig, ax1 = plt.subplots(figsize=(10,6))
    ax1.set_xlabel('Number of Topics K')
    ax1.set_ylabel('Coherence (c_v)', color='tab:blue')
    ax1.plot(ks, coherence_scores, marker='o', color='tab:blue', label='Coherence')
    ax1.tick_params(axis='y', labelcolor='tab:blue')

    ax2 = ax1.twinx()
    ax2.set_ylabel('Perplexity (log)', color='tab:red')
    ax2.plot(ks, perplexity_scores, marker='s', color='tab:red', label='Perplexity')
    ax2.tick_params(axis='y', labelcolor='tab:red')

    fig.suptitle('Optimal Number of Topics Selection')
    fig.legend(loc='upper right')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.savefig('optimal_k_plot.png', dpi=300)
    plt.show()
    print("最优主题数曲线已保存为 optimal_k_plot.png")


def get_topic_terms_distinctive(lda_model, dictionary, topn=10):
    """
    提取每个主题的 Top-n '独有'特征词。
    使用 c-TF-IDF 风格的区分度打分，抑制跨主题重复词。
    返回 topic_terms 列表，并保存 CSV。
    """
    num_topics = lda_model.num_topics
    vocab_size = len(dictionary)
    # 主题-词概率矩阵 (K x V)
    topic_term_matrix = lda_model.get_topics()

    # 统计每个词出现在多少个主题中（阈值 0.001 视为出现）
    presence_threshold = 1e-3
    word_presence = (topic_term_matrix > presence_threshold).sum(axis=0)  # shape (V,)
    word_presence = np.maximum(word_presence, 1)  # 避免除零

    # 计算区分度得分
    idf = np.log(num_topics / word_presence)  # shape (V,)
    distinctiveness = topic_term_matrix * idf[np.newaxis, :]  # 广播

    # 提取每个主题 top-n
    topic_terms = []
    print("\n=== 主题-术语表（按区分度排序）===\n")
    for idx in range(num_topics):
        # argsort 从大到小
        top_ids = np.argsort(distinctiveness[idx])[::-1][:topn]
        term_list = [(dictionary[i], round(float(distinctiveness[idx, i]), 4),
                      round(float(topic_term_matrix[idx, i]), 4)) for i in top_ids]
        topic_terms.append({'Topic_ID': idx, 'Terms': term_list})

        # 打印：术语(区分度分数, 原始概率)
        terms_str = ', '.join([f"{t}({s:.3f}|p={p:.3f})" for t, s, p in term_list])
        print(f"Topic {idx}: {terms_str}")

    # 保存 CSV
    rows = []
    for item in topic_terms:
        for term, score, prob in item['Terms']:
            rows.append({'Topic': item['Topic_ID'], 'Term': term,
                         'Distinctiveness': score, 'Probability': prob})
    pd.DataFrame(rows).to_csv('topic_terms_distinctive.csv',
                              index=False, encoding='utf-8-sig')
    print("\n主题术语已保存至 topic_terms_distinctive.csv")
    return topic_terms


def plot_topic_trends(df, lda_model, corpus, dictionary, num_topics):
    """
    计算每年每个主题的平均概率（强度），绘制时间趋势图。
    """
    # 为每篇文档推断主题概率
    doc_topics = []
    for bow in corpus:
        # 获取文档主题分布
        topics = lda_model.get_document_topics(bow, minimum_probability=0)
        topic_probs = [0] * num_topics
        for t, prob in topics:
            topic_probs[t] = prob
        doc_topics.append(topic_probs)
    df['topic_dist'] = doc_topics

    # 按年份聚合
    years = sorted(df['year'].unique())
    topic_mean_by_year = []
    for y in years:
        sub = df[df['year'] == y]
        if len(sub) == 0:
            continue
        # 计算该年所有文档的主题概率均值
        mean_probs = np.mean(np.vstack(sub['topic_dist'].values), axis=0)
        topic_mean_by_year.append([y] + list(mean_probs))
    df_trend = pd.DataFrame(topic_mean_by_year, columns=['Year'] + [f'T{i}' for i in range(num_topics)])

    # 绘制折线图
    plt.figure(figsize=(14, 8))
    for i in range(num_topics):
        plt.plot(df_trend['Year'], df_trend[f'T{i}'], marker='o', label=f'Topic {i}')
    plt.xlabel('Year')
    plt.ylabel('Average Topic Probability')
    plt.title('Topic Frequency Change Over Years')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.savefig('topic_trends.png', dpi=300)
    plt.show()
    print("主题趋势图已保存为 topic_trends.png")
    return df_trend


def export_vosviewer_network(lda_model, dictionary, threshold=0.3, filename='topic_network.net'):
    """
    基于主题-词项概率向量，计算主题间余弦相似度，生成 VOSviewer 网络文件。
    """
    num_topics = lda_model.num_topics
    # 获取主题-词项矩阵 (topics x vocab)
    topic_matrix = lda_model.get_topics()  # shape (num_topics, vocab_size)
    # 归一化
    normed = topic_matrix / np.linalg.norm(topic_matrix, axis=1, keepdims=True)
    sim_matrix = np.dot(normed, normed.T)  # 余弦相似度

    # 写入 .net 文件
    with open(filename, 'w') as f:
        f.write("*Vertices {}\n".format(num_topics))
        for i in range(num_topics):
            f.write('{} "Topic {}"\n'.format(i+1, i))
        f.write("*Edges\n")
        for i in range(num_topics):
            for j in range(i+1, num_topics):
                sim = sim_matrix[i, j]
                if sim > threshold:
                    f.write("{} {} {:.4f}\n".format(i+1, j+1, sim))
    print(f"VOSviewer 网络文件已保存为 {filename}")
    # 同时保存相似度矩阵供参考
    sim_df = pd.DataFrame(sim_matrix, index=[f'T{i}' for i in range(num_topics)],
                          columns=[f'T{i}' for i in range(num_topics)])
    sim_df.to_csv('topic_similarity_matrix.csv')
    print("主题相似度矩阵已保存为 topic_similarity_matrix.csv")


def main():
    np.random.seed(42)
    filepath = '/content/drive/MyDrive/savedrecs.xls'  # 请修改为你的实际路径

    # 1. 加载与预处理
    df, dictionary, corpus = load_and_preprocess(filepath)
    print(f"共处理 {len(df)} 篇文献，词典大小 {len(dictionary)}")

    # 2. 计算最优主题数（K 从 2 到 15）
    start_k, end_k = 2, 15
    models, coh_scores, perp_scores = compute_coherence_values(
        corpus, dictionary, df['tokens'].tolist(), start=start_k, end=end_k, step=1
    )
    plot_optimal_k(coh_scores, perp_scores, start_k, end_k)

    # 3. 选择最佳 K（此处以 coherence 最高为准）
    best_k = start_k + np.argmax(coh_scores)
    print(f"\n最佳主题数选择为 K = {best_k} (coherence={max(coh_scores):.4f})")
    lda_best = models[best_k]

    # 4. 输出主题-术语表
    topic_terms = get_topic_terms_distinctive(lda_best, dictionary, topn=10)

    # 5. 主题频率趋势图
    df_trend = plot_topic_trends(df, lda_best, corpus, dictionary, best_k)

    # 6. 导出 VOSviewer 网络文件
    export_vosviewer_network(lda_best, dictionary, threshold=0.3, filename='topic_network.net')

    print("\n分析完成！所有输出文件均已生成。")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python
# -*- coding: utf-8 -*-
"""
LDA 主题模型分析 - 滑坡文献数据
输入： savedrecs.xls（包含文献信息）
输出：
    - 最优主题数曲线（困惑度 & 一致性）
    - 主题-术语表（打印并保存为 CSV）
    - 各主题频率随年份变化趋势图（保存为 PNG）
    - 主题相似度矩阵及 VOSviewer 网络文件 (.net)
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import spacy
import gensim
from gensim.corpora import Dictionary
from gensim.models import LdaMulticore, LdaModel, CoherenceModel
from gensim.matutils import Sparse2Corpus
import warnings
import os
import logging

# 设置日志级别，屏蔽 gensim 的过多输出
logging.basicConfig(level=logging.ERROR)

# 下载 nltk 数据（首次运行时需取消注释）
# import nltk
# nltk.download('punkt')
# nltk.download('stopwords')

# 加载 spacy 英文模型（需先安装：python -m spacy download en_core_web_sm）
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])
# 扩展停用词
from nltk.corpus import stopwords

# ==================== 1. 数据读取与预处理 ====================
def load_and_preprocess(filepath):
    """
    读取 Excel/CSV 文件，合并文本字段，分词，构建词典和语料库。
    返回: (df, dictionary, corpus)
    """
    # ------ 1. 健壮读取文件 ------
    def read_file(fp):
        try:
            return pd.read_excel(fp)
        except Exception:
            try:
                return pd.read_csv(fp, encoding='utf-8')
            except Exception:
                try:
                    return pd.read_csv(fp, encoding='latin-1')
                except Exception:
                    return pd.read_csv(fp, sep='\t', encoding='utf-8')
    df = read_file(filepath)
    print("文件读取成功，列名：", df.columns.tolist())

    # ------ 2. 合并文本字段（根据实际列名调整）------
    # 注意：列名可能因文件格式而异，此处假设为原始列名，请根据实际调整
    text_cols = ['Article Title', 'Abstract', 'Author Keywords', 'Keywords Plus']
    # 检查列是否存在
    for col in text_cols:
        if col not in df.columns:
            print(f"警告：列 '{col}' 不存在，将跳过该列")
    df['combined'] = ''
    for col in text_cols:
        if col in df.columns:
            df['combined'] += df[col].fillna('') + ' '

    # ------ 3. 提取年份 ------
    if 'Publication Year' in df.columns:
        df['year'] = pd.to_numeric(df['Publication Year'], errors='coerce')
    else:
        # 尝试其他可能的年份列名
        year_col = next((c for c in df.columns if 'year' in c.lower()), None)
        if year_col:
            df['year'] = pd.to_numeric(df[year_col], errors='coerce')
        else:
            raise ValueError("未找到年份列，请检查数据")
    df = df.dropna(subset=['year'])
    df['year'] = df['year'].astype(int)
    extra_stopwords = {
    # 论文套话
    'study', 'paper', 'article', 'research', 'result', 'conclusion',
    'present', 'propose', 'proposed', 'show', 'shown', 'based', 'using',
    'used', 'use', 'also', 'however', 'therefore', 'may', 'can',
    # 泛化的量词/副词
    'high', 'low', 'large', 'small', 'different', 'various', 'significant',
    'important', 'main', 'major', 'new', 'recent',
    # 时间/地点通用词
    'year', 'time', 'day', 'period', 'region', 'area', 'site', 'location',
    # 泛化的名词
    'datum', 'base', 'factor', 'method', 'analysis', 'approach',
    'process', 'system', 'event', 'case', 'condition', 'characteristic',
    # 单位/格式词
    'et', 'al', 'fig', 'figure', 'table', 'eq', 'equation',
    }
    # 注意：不再包含 rainfall, trigger, flow, movement, debris, change, increase 等
    stop_words = set(stopwords.words('english')).union(extra_stopwords)
    # ------ 4. 文本预处理 ------
    def preprocess(text):
      if not isinstance(text, str):
          return []
      doc = nlp(text)
      tokens = [
          token.lemma_.lower()
          for token in doc
          if token.is_alpha
          and token.lemma_.lower() not in stop_words
          and token.pos_ in ('NOUN', 'ADJ', 'VERB')  # 恢复形容词和动词
          and len(token.lemma_) > 2  # 过滤过短的词
      ]
      return tokens

    print("正在预处理文本...")
    df['tokens'] = df['combined'].apply(preprocess)

    # 构建词典和语料库
    dictionary = Dictionary(df['tokens'])
    dictionary.filter_extremes(no_below=3, no_above=0.5)  # 从 0.6 降到 0.5，剔除更普遍的词
    corpus = [dictionary.doc2bow(tokens) for tokens in df['tokens']]

    return df, dictionary, corpus


def compute_coherence_values(corpus, dictionary, texts, start=2, end=15, step=1):
    """
    计算不同主题数 K 下的 Coherence (c_v) 和 Perplexity。
    返回各 K 下的 coherence 和 perplexity 列表。
    """
    coherence_scores = []
    perplexity_scores = []
    models = {}
    for k in range(start, end+1, step):
        print(f"训练 LDA K={k} ...")
        # 训练 LDA 模型，设定随机种子保证可重复
        lda = LdaModel(corpus=corpus, id2word=dictionary, num_topics=k,
                       passes=10, random_state=42, alpha='auto', eta='auto')
        models[k] = lda
        # 计算一致性
        cm = CoherenceModel(model=lda, texts=texts, dictionary=dictionary, coherence='c_v')
        coh = cm.get_coherence()
        # 计算困惑度
        perp = lda.log_perplexity(corpus)
        coherence_scores.append(coh)
        perplexity_scores.append(perp)
        print(f"K={k}   Coherence={coh:.4f}   Perplexity={perp:.4f}")
    return models, coherence_scores, perplexity_scores


def plot_optimal_k(coherence_scores, perplexity_scores, start_k, end_k):
    """
    绘制一致性（越高越好）和困惑度（越低越好）随 K 的变化曲线，辅助确定最优主题数。
    """
    ks = list(range(start_k, end_k+1))
    fig, ax1 = plt.subplots(figsize=(10,6))
    ax1.set_xlabel('Number of Topics K')
    ax1.set_ylabel('Coherence (c_v)', color='tab:blue')
    ax1.plot(ks, coherence_scores, marker='o', color='tab:blue', label='Coherence')
    ax1.tick_params(axis='y', labelcolor='tab:blue')

    ax2 = ax1.twinx()
    ax2.set_ylabel('Perplexity (log)', color='tab:red')
    ax2.plot(ks, perplexity_scores, marker='s', color='tab:red', label='Perplexity')
    ax2.tick_params(axis='y', labelcolor='tab:red')

    fig.suptitle('Optimal Number of Topics Selection')
    fig.legend(loc='upper right')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.savefig('optimal_k_plot.png', dpi=300)
    plt.show()
    print("最优主题数曲线已保存为 optimal_k_plot.png")


def get_topic_terms_exclusive(lda_model, dictionary, topn=10, max_topic_per_word=1):
    """
    提取每个主题的 Top-n 独有特征词，保证词项在主题间不重复。

    参数：
        topn: 每个主题列出的术语数量
        max_topic_per_word: 一个词最多可以出现在几个主题中
                            （1=完全互斥，2=允许出现在top2主题，默认1）

    返回：
        topic_terms: [{Topic_ID, Terms:[(term, score, prob), ...]}, ...]
    """
    num_topics = lda_model.num_topics
    vocab_size = len(dictionary)
    topic_term_matrix = lda_model.get_topics()  # (K, V)

    # ---------- 1. 计算区分度得分（c-TF-IDF 风格） ----------
    presence_threshold = 1e-3
    word_presence = (topic_term_matrix > presence_threshold).sum(axis=0)  # (V,)
    word_presence = np.maximum(word_presence, 1)
    idf = np.log(num_topics / word_presence)  # (V,)
    distinctiveness = topic_term_matrix * idf[np.newaxis, :]  # (K, V)

    # ---------- 2. 构建所有 (score, topic, word) 三元组并降序排列 ----------
    triples = []
    for t in range(num_topics):
        # 只考虑得分 > 0 的词，减少计算量
        valid_words = np.where(distinctiveness[t] > 0)[0]
        for w in valid_words:
            triples.append((distinctiveness[t, w], t, w))
    triples.sort(key=lambda x: x[0], reverse=True)

    # ---------- 3. 贪婪互斥分配 ----------
    topic_terms_ids = {t: [] for t in range(num_topics)}   # 每主题已分配的词 id
    word_assigned_count = {}                               # 每个词被分配了几次

    for score, t, w in triples:
        # 该主题已满
        if len(topic_terms_ids[t]) >= topn:
            continue
        # 该词已达到最大分配次数
        if word_assigned_count.get(w, 0) >= max_topic_per_word:
            continue
        topic_terms_ids[t].append(w)
        word_assigned_count[w] = word_assigned_count.get(w, 0) + 1

    # ---------- 4. 输出与保存 ----------
    print(f"\n=== 主题-术语表（互斥分配，每词最多出现 {max_topic_per_word} 次）===\n")
    topic_terms = []
    for t in range(num_topics):
        # 按区分度分数从高到低排序
        term_list = []
        for w in topic_terms_ids[t]:
            term = dictionary[w]
            score = float(distinctiveness[t, w])
            prob = float(topic_term_matrix[t, w])
            term_list.append((term, round(score, 4), round(prob, 4)))
        term_list.sort(key=lambda x: x[1], reverse=True)

        topic_terms.append({'Topic_ID': t, 'Terms': term_list})

        terms_str = ', '.join([f"{term}({score:.3f}|p={prob:.3f})"
                               for term, score, prob in term_list])
        print(f"Topic {t}: {terms_str}")

    # 保存 CSV
    rows = []
    for item in topic_terms:
        for term, score, prob in item['Terms']:
            rows.append({'Topic': item['Topic_ID'], 'Term': term,
                         'Distinctiveness': score, 'Probability': prob})
    pd.DataFrame(rows).to_csv('topic_terms_exclusive.csv',
                              index=False, encoding='utf-8-sig')
    print("\n互斥主题术语已保存至 topic_terms_exclusive.csv")
    return topic_terms


def plot_topic_trends(df, lda_model, corpus, dictionary, num_topics=None,
                      window=3, smooth_only=True):
    """
    计算每年每个主题的平均概率，并绘制趋势图（含3年滑动平均）。

    参数：
        window: 滑动窗口大小，默认 3 年
        smooth_only: True 只画平滑曲线；False 同时画原始曲线（淡色）
    """
    if num_topics is None:
        num_topics = lda_model.num_topics
    else:
        if num_topics != lda_model.num_topics:
            num_topics = lda_model.num_topics

    # ---------- 1. 推断每篇文档的主题分布 ----------
    doc_topics = []
    for bow in corpus:
        topics = lda_model.get_document_topics(bow, minimum_probability=0)
        probs = [0.0] * num_topics
        for t, p in topics:
            if t < num_topics:
                probs[t] = p
        doc_topics.append(probs)
    df = df.copy()
    df['topic_dist'] = doc_topics

    # ---------- 2. 按年份聚合并排序 ----------
    years = sorted(df['year'].unique())
    rows = []
    for y in years:
        sub = df[df['year'] == y]
        if len(sub) == 0:
            continue
        mean_probs = np.mean(np.vstack(sub['topic_dist'].values), axis=0)
        rows.append([y] + list(mean_probs))
    df_trend = pd.DataFrame(rows, columns=['Year'] + [f'T{i}' for i in range(num_topics)])
    df_trend = df_trend.sort_values('Year').reset_index(drop=True)

    # ---------- 3. 3年滑动平均 ----------
    df_smooth = df_trend.copy()
    for i in range(num_topics):
        df_smooth[f'T{i}'] = (df_trend[f'T{i}']
                              .rolling(window=window, center=True, min_periods=1)
                              .mean())

    # ---------- 4. 绘图 ----------
    plt.figure(figsize=(14, 8))
    cmap = plt.cm.tab10
    for i in range(num_topics):
        color = cmap(i % 10)
        if not smooth_only:
            # 淡色原始曲线 + 半透明
            plt.plot(df_trend['Year'], df_trend[f'T{i}'],
                     color=color, alpha=0.25, linewidth=1,
                     marker='o', markersize=3)
        # 加粗平滑曲线
        plt.plot(df_smooth['Year'], df_smooth[f'T{i}'],
                 color=color, linewidth=2.2, marker='o', markersize=5,
                 label=f'Topic {i}')

    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Average Topic Probability (3-year moving average)', fontsize=12)
    plt.title('Topic Frequency Change Over Years (3-year moving average)', fontsize=14)
    plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=10)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.savefig('topic_trends_smooth.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("主题趋势图（3年滑动平均）已保存为 topic_trends_smooth.png")

    # 同时保存平滑后的数据
    df_smooth.to_csv('topic_trends_smooth.csv', index=False, encoding='utf-8-sig')
    print("平滑后的趋势数据已保存为 topic_trends_smooth.csv")

    return df_trend, df_smooth


def export_vosviewer_network(lda_model, dictionary, threshold=0.3, filename='topic_network.net'):
    """
    基于主题-词项概率向量，计算主题间余弦相似度，生成 VOSviewer 网络文件。
    """
    num_topics = lda_model.num_topics
    # 获取主题-词项矩阵 (topics x vocab)
    topic_matrix = lda_model.get_topics()  # shape (num_topics, vocab_size)
    # 归一化
    normed = topic_matrix / np.linalg.norm(topic_matrix, axis=1, keepdims=True)
    sim_matrix = np.dot(normed, normed.T)  # 余弦相似度

    # 写入 .net 文件
    with open(filename, 'w') as f:
        f.write("*Vertices {}\n".format(num_topics))
        for i in range(num_topics):
            f.write('{} "Topic {}"\n'.format(i+1, i))
        f.write("*Edges\n")
        for i in range(num_topics):
            for j in range(i+1, num_topics):
                sim = sim_matrix[i, j]
                if sim > threshold:
                    f.write("{} {} {:.4f}\n".format(i+1, j+1, sim))
    print(f"VOSviewer 网络文件已保存为 {filename}")
    # 同时保存相似度矩阵供参考
    sim_df = pd.DataFrame(sim_matrix, index=[f'T{i}' for i in range(num_topics)],
                          columns=[f'T{i}' for i in range(num_topics)])
    sim_df.to_csv('topic_similarity_matrix.csv')
    print("主题相似度矩阵已保存为 topic_similarity_matrix.csv")


def main():
    np.random.seed(42)
    filepath = '/content/drive/MyDrive/savedrecs.xls'  # 请修改为你的实际路径

    # 1. 加载与预处理
    df, dictionary, corpus = load_and_preprocess(filepath)
    print(f"共处理 {len(df)} 篇文献，词典大小 {len(dictionary)}")

    # 2. 计算最优主题数（K 从 2 到 15）
    start_k, end_k = 2, 15
    models, coh_scores, perp_scores = compute_coherence_values(
        corpus, dictionary, df['tokens'].tolist(), start=start_k, end=end_k, step=1
    )
    plot_optimal_k(coh_scores, perp_scores, start_k, end_k)

    # 3. 选择最佳 K（此处以 coherence 最高为准）
    best_k = start_k + np.argmax(coh_scores)
    print(f"\n最佳主题数选择为 K = {best_k} (coherence={max(coh_scores):.4f})")
    lda_best = models[best_k]

    # 4. 输出主题-术语表
    topic_terms = get_topic_terms_exclusive(
        lda_best, dictionary,
        topn=10,
        max_topic_per_word=1   # 1=完全互斥；2=允许重复出现在最多2个主题
    )

    # 5. 主题频率趋势图
    df_trend, df_smooth = plot_topic_trends(
        df, lda_best, corpus, dictionary,
        num_topics=best_k,
        window=3,           # 3年滑动平均
        smooth_only=True    # 只显示平滑曲线；想同时显示原始曲线改成 False
    )

    # 6. 导出 VOSviewer 网络文件
    export_vosviewer_network(lda_best, dictionary, threshold=0.3, filename='topic_network.net')

    print("\n分析完成！所有输出文件均已生成。")

if __name__ == "__main__":
    main()
```
