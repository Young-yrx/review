```python
# -*- coding: utf-8 -*-
"""
极端降雨群发滑坡智能识别研究 - 发文量趋势分析
基于真实 WoS 数据 (savedrecs.xls)
完整绘图代码
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import gridspec
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
# ==================== 1. 设置中文字体 ====================
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 120
```

```python
# ==================== 2. 加载真实数据 ====================
# 读取 WoS 导出文件（请确保 savedrecs.xls 位于当前工作目录）
df_raw = pd.read_excel('/content/drive/MyDrive/savedrecs.xls', engine='xlrd')

# 检查年份列名称（常见为 'Publication Year' 或 'PY'）
year_col = 'Publication Year' if 'Publication Year' in df_raw.columns else 'PY'
if year_col not in df_raw.columns:
    raise KeyError("未找到年份列，请检查数据列名。")

# 提取年份并统计发文量
year_series = df_raw[year_col].dropna().astype(int)
year_counts = year_series.value_counts().sort_index()

# 生成完整的年份范围（从最早到最晚，补全缺失年份）
all_years = np.arange(year_counts.index.min(), year_counts.index.max() + 1)
counts = np.array([year_counts.get(y, 0) for y in all_years])

# 构建 DataFrame
df = pd.DataFrame({'year': all_years, 'count': counts})
df['cumulative'] = df['count'].cumsum()

print(f"数据年份范围：{df['year'].min()} - {df['year'].max()}")
print(f"总发文量：{df['count'].sum()} 篇")
print(f"总年份数：{len(df)} 年")
```

```python
# ==================== 3. 阶段划分（根据实际趋势手动调整） ====================
stages = [
    ('Emergence Stage', 2002, 2012),
    ('Steady Growth Stage', 2013, 2019),
    ('Rapid Growth Stage', 2020, 2025)
]
```

```python
# ==================== 4. 重大极端降雨事件（根据事实标注） ====================
events = [
    (2021, 'Henan "7·20" Rainstorm'),
    (2022, 'Guangxi Longsheng Landslide'),
    (2023, 'Typhoon Haikui'),
    (2024, 'Chongqing Pengshui Rainstorm'),

    # 新增重大事件（建议选取 5-6 个，请按需调整）
    (2013, 'Sichuan Lushan Rain-induced Slide'),  # 四川芦山地震后强降雨引发滑坡（可选）
    (2017, 'Sichuan Maoxian Landslide'),          # 四川茂县叠溪镇山体滑坡（2017.6.24））
]
```

```python
# ==================== 5. 多项式拟合（累计发文量） ====================
X = df['year'].values.reshape(-1, 1)
y = df['cumulative'].values
poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)
model = LinearRegression()
model.fit(X_poly, y)
y_pred = model.predict(X_poly)
r2 = r2_score(y, y_pred)

# 提取系数用于显示方程
coeffs = model.coef_
intercept = model.intercept_

```

```python
# ==================== 6. 绘制组合图 ====================
fig = plt.figure(figsize=(14, 10))
gs = gridspec.GridSpec(2, 1, height_ratios=[3, 2], hspace=0.35)

# ---------- 子图1：年度发文量 ----------
ax1 = fig.add_subplot(gs[0])

# 主折线
ax1.plot(df['year'], df['count'], 'o-', color='#2c3e50',
         linewidth=2.5, markersize=7, label='Annual Publications', zorder=2)

# 阶段划分（垂直虚线 + 文本标签）
stage_colors = ['#95a5a6', '#3498db', '#e67e22']
for i, (name, start, end) in enumerate(stages):
    mid_year = (start + end) // 2
    ax1.axvline(x=start, color=stage_colors[i], linestyle='--',
                alpha=0.6, linewidth=1.5, zorder=1)
    # 阶段名称标注在图表顶部
    ax1.text(mid_year, ax1.get_ylim()[1] * 0.93, name,
             ha='center', fontsize=13, color=stage_colors[i],
             weight='bold', zorder=3)

# 重大事件标注（红色三角在顶部，并旋转文本）
for year, label in events:
    if year in df['year'].values:
        # 获取该年的发文量
        count_val = df.loc[df['year'] == year, 'count'].iloc[0]
        # 在数据点位置画红色星号或三角形（更醒目）
        ax1.scatter(year, count_val, marker='^', s=120,
                    color='#c0392b', zorder=3, edgecolor='white', linewidth=1)
        # 在点上方加英文标签（向上偏移一点避免重叠）
        ax1.text(year, count_val + 2, label,
                 ha='center', va='bottom', fontsize=8,
                 color='#c0392b', weight='bold', zorder=3)

# 坐标轴设置
ax1.set_ylabel('Number of Publications', fontsize=13)
ax1.set_xlim(df['year'].min() - 2, df['year'].max() + 2)
ax1.set_title('Annual Publication Trends in Extreme Rainfall-Induced Landslide Research', fontsize=15, weight='bold')
ax1.grid(True, linestyle=':', alpha=0.5, axis='y')
ax1.legend(loc='lower right', fontsize=11)
ax1.tick_params(labelsize=11)
# （可选）在折线上显示数值，若年份太多可注释掉
# for x, y_val in zip(df['year'], df['count']):
#     if y_val > 0:
#         ax1.text(x, y_val + 2, str(y_val), ha='center', fontsize=8)

# ---------- 子图2：累计发文量与拟合曲线 ----------
ax2 = fig.add_subplot(gs[1])

# 实际累计曲线
ax2.plot(df['year'], df['cumulative'], 'o-', color='#2c3e50',
         linewidth=2, markersize=5, label='Cumulative Publications', zorder=2)

# 三阶多项式拟合曲线（红色虚线）
ax2.plot(df['year'], y_pred, '--', color='#e74c3c',
         linewidth=2.5, label=f'Cubic Polynomial Fit (R² = {r2:.4f})', zorder=3)

# 显示拟合方程
eq_text = (f'y = {coeffs[3]:.3f}x³ + {coeffs[2]:.3f}x² + '
           f'{coeffs[1]:.3f}x + {intercept:.0f}')
ax2.text(0.05, 0.92, eq_text, transform=ax2.transAxes,
         fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='#fef9e7', edgecolor='#e74c3c', alpha=0.9))

# 坐标轴设置
ax2.set_xlabel('Year', fontsize=13)
ax2.set_ylabel('Cumulative Publications', fontsize=13)
ax2.set_xlim(df['year'].min() - 2, df['year'].max() + 2)
ax2.grid(True, linestyle=':', alpha=0.5, axis='y')
ax2.legend(loc='lower right', fontsize=11)   # 位置可根据需要调整，此处保持原位置
ax2.tick_params(labelsize=11)
```

```python
# ==================== 7. 保存与显示 ====================
plt.tight_layout()
plt.savefig('发文量趋势图.png', dpi=300, bbox_inches='tight')
plt.show()

print(f"\n多项式拟合 R² = {r2:.4f}")
print("图形已保存为 发文量趋势图.png")
```
