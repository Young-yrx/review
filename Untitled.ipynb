#文章分析

```python
# -*- coding: utf-8 -*-
"""
极端降雨群发滑坡智能识别研究 - 发文量趋势分析
基于真实 WoS 数据 (savedrecs.xls)
完整绘图代码
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import gridspec
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from collections import defaultdict, Counter
import math
import re

```

```python
# ==================== 1. 设置中文字体 ====================
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 120
```

```python
# ==================== 2. 加载真实数据（基于 Early Access Date） ====================
df_raw = pd.read_excel('/content/drive/MyDrive/savedrecs.xls', engine='xlrd')

date_col = 'Early Access Date'
if date_col not in df_raw.columns:
    raise KeyError(f"未找到 '{date_col}' 列，请检查数据列名。")

def extract_year(date_str):
    if pd.isna(date_str):
        return np.nan
    s = str(date_str).strip()
    if not s:
        return np.nan
    parts = s.split()
    try:
        return int(parts[-1])
    except ValueError:
        try:
            return int(s)
        except ValueError:
            return np.nan

df_raw['year_extracted'] = df_raw[date_col].apply(extract_year)

year_series = df_raw['year_extracted'].dropna().astype(int)
year_counts = year_series.value_counts().sort_index()

all_years = np.arange(year_counts.index.min(), year_counts.index.max() + 1)
counts = np.array([year_counts.get(y, 0) for y in all_years])

df = pd.DataFrame({'year': all_years, 'count': counts})
df['cumulative'] = df['count'].cumsum()

print(f"数据年份范围：{df['year'].min()} - {df['year'].max()}")
print(f"总发文量：{df['count'].sum()} 篇")
print(f"总年份数：{len(df)} 年")
```

```python
# ==================== 3. 阶段划分（根据实际趋势手动调整） ====================
stages = [
    ('Emergence Stage', 2002, 2012),
    ('Steady Growth Stage', 2013, 2019),
    ('Rapid Growth Stage', 2020, 2025)
]
```

```python
# ==================== 4. 重大极端降雨事件（根据事实标注） ====================
events = [
    (2021, 'Henan "7·20" Rainstorm'),
    (2022, 'Guangxi Longsheng Landslide'),
    (2023, 'Typhoon Haikui'),
    (2024, 'Chongqing Pengshui Rainstorm'),

    # 新增重大事件（建议选取 5-6 个，请按需调整）
    (2013, 'Sichuan Lushan Rain-induced Slide'),  # 四川芦山地震后强降雨引发滑坡（可选）
    (2017, 'Sichuan Maoxian Landslide'),          # 四川茂县叠溪镇山体滑坡（2017.6.24））
]
```

```python
# ==================== 5. 多项式拟合（累计发文量） ====================
X = df['year'].values.reshape(-1, 1)
y = df['cumulative'].values
poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)
model = LinearRegression()
model.fit(X_poly, y)
y_pred = model.predict(X_poly)
r2 = r2_score(y, y_pred)

# 提取系数用于显示方程
coeffs = model.coef_
intercept = model.intercept_

```

```python
# ==================== 6. 绘制组合图 ====================
fig = plt.figure(figsize=(14, 10))
gs = gridspec.GridSpec(2, 1, height_ratios=[3, 2], hspace=0.35)

# ---------- 子图1：年度发文量 ----------
ax1 = fig.add_subplot(gs[0])

# 主折线
ax1.plot(df['year'], df['count'], 'o-', color='#2c3e50',
         linewidth=2.5, markersize=7, label='Annual Publications', zorder=2)

# 阶段划分（垂直虚线 + 文本标签）
stage_colors = ['#95a5a6', '#3498db', '#e67e22']
for i, (name, start, end) in enumerate(stages):
    mid_year = (start + end) // 2
    ax1.axvline(x=start, color=stage_colors[i], linestyle='--',
                alpha=0.6, linewidth=1.5, zorder=1)
    # 阶段名称标注在图表顶部
    ax1.text(mid_year, ax1.get_ylim()[1] * 0.93, name,
             ha='center', fontsize=13, color=stage_colors[i],
             weight='bold', zorder=3)

# 重大事件标注（红色三角在顶部，并旋转文本）
for year, label in events:
    if year in df['year'].values:
        # 获取该年的发文量
        count_val = df.loc[df['year'] == year, 'count'].iloc[0]
        # 在数据点位置画红色星号或三角形（更醒目）
        ax1.scatter(year, count_val, marker='^', s=120,
                    color='#c0392b', zorder=3, edgecolor='white', linewidth=1)
        # 在点上方加英文标签（向上偏移一点避免重叠）
        ax1.text(year, count_val + 2, label,
                 ha='center', va='bottom', fontsize=8,
                 color='#c0392b', weight='bold', zorder=3)

# 坐标轴设置
ax1.set_ylabel('Number of Publications', fontsize=13)
ax1.set_xlim(df['year'].min() - 2, df['year'].max() + 2)
ax1.set_title('Annual Publication Trends in Extreme Rainfall-Induced Landslide Research', fontsize=15, weight='bold')
ax1.grid(True, linestyle=':', alpha=0.5, axis='y')
ax1.legend(loc='lower right', fontsize=11)
ax1.tick_params(labelsize=11)
# （可选）在折线上显示数值，若年份太多可注释掉
# for x, y_val in zip(df['year'], df['count']):
#     if y_val > 0:
#         ax1.text(x, y_val + 2, str(y_val), ha='center', fontsize=8)

# ---------- 子图2：累计发文量与拟合曲线 ----------
ax2 = fig.add_subplot(gs[1])

# 实际累计曲线
ax2.plot(df['year'], df['cumulative'], 'o-', color='#2c3e50',
         linewidth=2, markersize=5, label='Cumulative Publications', zorder=2)

# 三阶多项式拟合曲线（红色虚线）
ax2.plot(df['year'], y_pred, '--', color='#e74c3c',
         linewidth=2.5, label=f'Cubic Polynomial Fit (R² = {r2:.4f})', zorder=3)

# 显示拟合方程
eq_text = (f'y = {coeffs[3]:.3f}x³ + {coeffs[2]:.3f}x² + '
           f'{coeffs[1]:.3f}x + {intercept:.0f}')
ax2.text(0.05, 0.92, eq_text, transform=ax2.transAxes,
         fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='#fef9e7', edgecolor='#e74c3c', alpha=0.9))

# 坐标轴设置
ax2.set_xlabel('Year', fontsize=13)
ax2.set_ylabel('Cumulative Publications', fontsize=13)
ax2.set_xlim(df['year'].min() - 2, df['year'].max() + 2)
ax2.grid(True, linestyle=':', alpha=0.5, axis='y')
ax2.legend(loc='lower right', fontsize=11)
ax2.tick_params(labelsize=11)
```

```python
# ==================== 7. 保存与显示 ====================
plt.tight_layout()
plt.savefig('发文量趋势图.png', dpi=300, bbox_inches='tight')
plt.show()

print(f"\n多项式拟合 R² = {r2:.4f}")
print("图形已保存为 发文量趋势图.png")
```

#作者分析

```python
# ------------------------------
# 1. 读取数据
# ------------------------------
file_path = '/content/drive/MyDrive/savedrecs.xls'  # 请根据实际路径修改
df = pd.read_excel(file_path, sheet_name=0)  # 读取第一个sheet

# 检查必要的列是否存在
required_cols = ['Authors', 'Times Cited, WoS Core']
for col in required_cols:
    if col not in df.columns:
        raise ValueError(f"缺少列: {col}")

# 处理被引次数，转为数值，缺失值填0
df['Times Cited, WoS Core'] = pd.to_numeric(df['Times Cited, WoS Core'], errors='coerce').fillna(0)

# 只保留有作者的行
df = df[df['Authors'].notna()]
```

```python
# ------------------------------
# 2. 解析作者名单（支持分号分隔）
# ------------------------------
# 作者格式示例: "Amankwah, SOY; Wang, GJ; Gnyawali, K; ..."
# 为了统一，去除多余空格，并拆分为列表
df['Authors_parsed'] = df['Authors'].apply(lambda x: [name.strip() for name in x.split(';') if name.strip()])
```

```python
# ------------------------------
# 3. 构建作者-论文映射
# ------------------------------
# 记录每篇文章的被引次数（用于后续计算）
papers = df[['Authors_parsed', 'Times Cited, WoS Core']].copy()
```

```python
# ------------------------------
# 4. 计算作者指标
# ------------------------------
# 4.1 总发文量 & 总被引
author_pub_count = defaultdict(int)
author_citation_sum = defaultdict(float)

for idx, row in papers.iterrows():
    authors = row['Authors_parsed']
    citations = row['Times Cited, WoS Core']
    for author in authors:
        author_pub_count[author] += 1
        author_citation_sum[author] += citations

# 4.2 篇均被引
author_cited_per_pub = {a: author_citation_sum[a] / author_pub_count[a] for a in author_pub_count}

# 4.3 H指数
def compute_h_index(citation_list):
    citation_list = sorted([c for c in citation_list if c > 0], reverse=True)
    h = 0
    for i, c in enumerate(citation_list, 1):
        if c >= i:
            h = i
        else:
            break
    return h

# 收集每位作者的所有单篇被引次数
author_citations_list = defaultdict(list)
for idx, row in papers.iterrows():
    citations = row['Times Cited, WoS Core']
    for author in row['Authors_parsed']:
        if citations > 0:
            author_citations_list[author].append(citations)

author_h_index = {a: compute_h_index(author_citations_list[a]) for a in author_pub_count}

# 4.4 合作分析
# 构建共现矩阵：统计每对作者合作次数
co_author_counts = defaultdict(Counter)

for idx, row in papers.iterrows():
    authors = row['Authors_parsed']
    for i, a1 in enumerate(authors):
        for a2 in authors[i+1:]:
            if a1 != a2:
                co_author_counts[a1][a2] += 1
                co_author_counts[a2][a1] += 1

# 4.5 最紧密合作对象 & 合作总数
most_frequent_coauthor = {}
total_coauthors = {}

for author in author_pub_count.keys():
    if author in co_author_counts:
        # 合作总数：不重复合作者数量
        total_coauthors[author] = len(co_author_counts[author])
        # 最紧密合作对象（合作次数最多）
        if co_author_counts[author]:
            most_freq = max(co_author_counts[author].items(), key=lambda x: x[1])
            most_frequent_coauthor[author] = most_freq[0]  # 只存名字，也可存次数
        else:
            most_frequent_coauthor[author] = None
    else:
        total_coauthors[author] = 0
        most_frequent_coauthor[author] = None
```

```python

# ------------------------------
# 5. 构建作者信息DataFrame
# ------------------------------
author_df = pd.DataFrame({
    'Author': list(author_pub_count.keys()),
    '总发文量': list(author_pub_count.values()),
    '总被引': list(author_citation_sum.values()),
    '篇均被引': [author_cited_per_pub[a] for a in author_pub_count.keys()],
    'H指数': [author_h_index[a] for a in author_pub_count.keys()],
    '合作总数': [total_coauthors.get(a, 0) for a in author_pub_count.keys()],
    '最紧密合作对象': [most_frequent_coauthor.get(a, '') for a in author_pub_count.keys()]
})

# 按总发文量降序排序
author_df_sorted = author_df.sort_values('总发文量', ascending=False).reset_index(drop=True)
```

```python
# ------------------------------
# 6. 输出结果
# ------------------------------
print("核心作者排名表（按总发文量）：")
print(author_df_sorted.to_string(index=False))

# 保存为Excel文件
author_df_sorted.to_excel("核心作者分析结果.xlsx", index=False)
```

#机构分析

```python


# 检查必要列是否存在
required_cols = ['Authors', 'Affiliations', 'Times Cited, WoS Core']
for col in required_cols:
    if col not in df.columns:
        raise KeyError(f"列 '{col}' 不存在，请检查数据列名。")

# 处理缺失值
df['Authors'] = df['Authors'].fillna('')
df['Affiliations'] = df['Affiliations'].fillna('')
df['Times Cited, WoS Core'] = df['Times Cited, WoS Core'].fillna(0).astype(int)

# 定义机构合并函数（中国科学院）
def normalize_institution(inst):
    """
    将机构名称标准化，合并中国科学院及其子机构为 'Chinese Academy of Sciences'
    """
    if not isinstance(inst, str):
        return inst
    # 1. 合并中国科学院及子机构
    if any(key in inst for key in ['Chinese Acad Sci', 'Chinese Academy of Sciences', 'Institute of Mountain Hazards', 'Inst Mt Hazards']):
        return 'Chinese Academy of Sciences'
    # 2. 合并中国地质科学院及子机构
    if any(key in inst for key in ['Chinese Acad Geol Sci', 'Chinese Academy of Geological Sciences', 'Institute of Geomechanics']):
        return 'Chinese Academy of Geological Sciences'
    # 3. 合并印度理工学院及所有分校
    if 'Indian Institute of Technology' in inst:
        return 'Indian Institute of Technology System (IIT System)'
    # 4. 合并印度空间研究组织与空间部
    if 'Indian Space Research Organisation' in inst or 'Department of Space' in inst:
        return 'Indian Space Research Organisation (ISRO)'

    return inst

# 解析作者和机构列表
def parse_authors(authors_str):
    """解析作者字符串，返回作者列表"""
    if pd.isna(authors_str) or authors_str == '':
        return []
    # 按分号分割，并去除前后空格
    authors = [a.strip() for a in authors_str.split(';') if a.strip()]
    return authors

def parse_affiliations(aff_str):
    """解析机构字符串，返回机构列表（标准化后）"""
    if pd.isna(aff_str) or aff_str == '':
        return []
    # 按分号分割
    insts = [i.strip() for i in aff_str.split(';') if i.strip()]
    # 标准化每个机构
    insts = [normalize_institution(i) for i in insts]
    return insts

# 用于存储每个机构的论文信息
paper_info = defaultdict(list)  # 机构 -> [(times_cited, authors_list, insts_list)]

# 逐行处理
for idx, row in df.iterrows():
    authors_str = row['Authors']
    aff_str = row['Affiliations']
    cited = row['Times Cited, WoS Core']

    authors = parse_authors(authors_str)
    insts = parse_affiliations(aff_str)

    if not insts:  # 无机构信息则跳过
        continue

    # 对每个机构记录该论文的信息
    # 注意：一篇论文可能对应多个机构，每个机构都获得该论文的贡献
    for inst in set(insts):  # 去重，避免同一机构重复计数
        paper_info[inst].append({
            'cited': cited,
            'authors': authors,
            'insts': insts  # 用于合作机构分析
        })

# 计算各机构的统计指标
results = []
for inst, papers in paper_info.items():
    total_pubs = len(papers)
    total_citations = sum(p['cited'] for p in papers)
    avg_citations = total_citations / total_pubs if total_pubs > 0 else 0

    # H指数
    citations_list = sorted([p['cited'] for p in papers], reverse=True)
    h_index = 0
    for i, c in enumerate(citations_list, 1):
        if c >= i:
            h_index = i
        else:
            break

    # 合作机构总数：统计该机构所有论文中出现的其他机构（去重）
    co_insts = set()
    for p in papers:
        for other_inst in p['insts']:
            if other_inst != inst:
                co_insts.add(other_inst)
    co_count = len(co_insts)

    # 研究人员数：统计该机构所有论文中的所有作者（去重）
    # 这里采用简单方法：将论文的所有作者都计入该机构，但可能导致作者跨机构重复，但作为近似
    # 更精确应基于作者-机构对应关系，但此处简化。
    researchers = set()
    for p in papers:
        for author in p['authors']:
            researchers.add(author)
    researcher_count = len(researchers)

    # 人均产出
    per_capita = total_pubs / researcher_count if researcher_count > 0 else 0

    results.append({
        '机构': inst,
        '总发文量': total_pubs,
        '总被引': total_citations,
        '篇均被引': round(avg_citations, 2),
        'H指数': h_index,
        '合作机构总数': co_count,
        '研究人员数': researcher_count,
        '人均产出': round(per_capita, 3)
    })

# 转换为DataFrame并排序（按总发文量降序）
df_results = pd.DataFrame(results)
df_results = df_results.sort_values('总发文量', ascending=False).reset_index(drop=True)

# 输出前20名核心机构
print("核心机构排名（按发文量）")
print(df_results.head(20).to_string(index=False))

# 可保存为Excel
df_results.to_excel('institution_analysis.xlsx', index=False)
```

#国家分析

```python


# 国家名称映射
country_mapping = {
    'Peoples R China': 'China',
    'USA': 'United States',
    'England': 'UK',
    'Scotland': 'UK',
    'Wales': 'UK',
    'South Korea': 'Korea',
    # 根据需要补充
}

def extract_countries(addr_str):
    """从地址字符串中提取所有出现的不重复国家"""
    if pd.isna(addr_str):
        return []
    countries = set()
    # 以分号分隔各个地址块
    for addr in addr_str.split(';'):
        # 取最后一个逗号后的部分作为国家
        parts = addr.split(',')
        if len(parts) > 1:
            country = parts[-1].strip()
            # 简单清洗：去除末尾句号或括号等
            country = re.sub(r'[\.\)].*$', '', country).strip()
            if country in country_mapping:
                country = country_mapping[country]
            if country:  # 非空
                countries.add(country)
    return list(countries)

# 提取每篇论文的国家列表
df['countries'] = df['Addresses'].apply(extract_countries)

# 删除无国家信息的记录（可选）
df = df[df['countries'].map(len) > 0]

# 获取被引次数（若缺失则按0计）
df['cites'] = pd.to_numeric(df['Times Cited, WoS Core'], errors='coerce').fillna(0)

# ---------- 国家统计 ----------
nation_papers = defaultdict(list)  # 国家 -> 该国家所有论文的被引次数列表

for idx, row in df.iterrows():
    for c in row['countries']:
        nation_papers[c].append(row['cites'])

stats = {}
for c, cites_list in nation_papers.items():
    n = len(cites_list)
    total_cites = sum(cites_list)
    avg_cites = total_cites / n if n > 0 else 0
    # 计算H指数
    cites_sorted = sorted(cites_list, reverse=True)
    h = 0
    for i, c_val in enumerate(cites_sorted, 1):
        if c_val >= i:
            h = i
        else:
            break
    stats[c] = {
        'papers': n,
        'total_cites': total_cites,
        'avg_cites': round(avg_cites, 2),
        'h_index': h
    }

# ---------- 合作网络 ----------
coop = defaultdict(Counter)  # 国家A -> Counter({国家B: 合作次数})
for idx, row in df.iterrows():
    cs = row['countries']
    if len(cs) > 1:
        for i in range(len(cs)):
            for j in range(i+1, len(cs)):
                coop[cs[i]][cs[j]] += 1
                coop[cs[j]][cs[i]] += 1

# 最紧密合作对象
most_coop = {}
for c in stats:
    if c in coop:
        most = coop[c].most_common(1)
        if most:
            most_coop[c] = most[0][0]

# 输出结果（按发文量降序）
print("国家\t发文量\t总被引\t篇均被引\tH指数\t最紧密合作对象\t合作国家数")
for c in sorted(stats, key=lambda x: stats[x]['papers'], reverse=True):
    data = stats[c]
    partner = most_coop.get(c, '无')
    num_partners = len(coop.get(c, {}))
    print(f"{c}\t{data['papers']}\t{data['total_cites']}\t{data['avg_cites']}\t{data['h_index']}\t{partner}\t{num_partners}")
```

##统计指标计算

```python
def compute_h_index(citations_list):
    """计算H指数"""
    citations = sorted(citations_list, reverse=True)
    h = 0
    for i, c in enumerate(citations, start=1):
        if c >= i:
            h = i
        else:
            break
    return h

# 按期刊分组
grouped = df_clean.groupby(journal_col)

journal_stats = {}
for journal, group in grouped:
    n = len(group)
    total_cites = group[cited_col].sum()
    avg_cites = total_cites / n if n > 0 else 0
    h_index = compute_h_index(group[cited_col].tolist())
    journal_stats[journal] = {
        '发文量': n,
        '总被引': total_cites,
        '篇均被引': avg_cites,
        'H指数': h_index
    }

# 转为DataFrame便于排序
stats_df = pd.DataFrame.from_dict(journal_stats, orient='index')
stats_df.index.name = '期刊'
stats_df.reset_index(inplace=True)

# 按总发文量降序排序
stats_df_sorted = stats_df.sort_values('发文量', ascending=False)
```

##期刊排名表（核心期刊）

```python
print(stats_df_sorted)
```

##核心期刊群识别

```python
# 计算累积发文量占比
stats_df_sorted['累积发文量'] = stats_df_sorted['发文量'].cumsum()
total_papers = stats_df_sorted['发文量'].sum()
stats_df_sorted['累积占比'] = stats_df_sorted['累积发文量'] / total_papers * 100

# 找出累积占比首次超过80%的位置
core_num = (stats_df_sorted['累积占比'] <= 80).sum() + 1
core_journals = stats_df_sorted.head(core_num)

print(f"核心期刊数量（累积发文量>80%）：{core_num}")
print(core_journals[['期刊', '发文量', '累积占比']])
```

##学科偏好分类

```python
# 手动构建分类字典（基于期刊名称关键词）
def classify_journal(name):
    name_lower = name.lower()
    if any(k in name_lower for k in ['eng geol', 'landslide', 'georisk', 'soildyn']):
        return '工程地质'
    elif any(k in name_lower for k in ['geomorph', 'catena', 'earth surf', 'mountain sci']):
        return '地貌'
    elif any(k in name_lower for k in ['remote sens', 'isprs', 'geoinf', 'gis']):
        return '遥感'
    else:
        return '综合'

stats_df['学科分类'] = stats_df['期刊'].apply(classify_journal)
group_by_category = stats_df.groupby('学科分类').agg({
    '发文量': 'sum',
    '总被引': 'sum',
    '篇均被引': 'mean'
}).round(2)

print(group_by_category)
```
